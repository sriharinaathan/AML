```python
import random
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from collections import defaultdict
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report
)
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.keras.utils.set_random_seed(SEED)
LINES = [
    (0, 1, 2),
    (3, 4, 5),
    (6, 7, 8),
    (0, 3, 6),
    (1, 4, 7),
    (2, 5, 8),
    (0, 4, 8),
    (2, 4, 6)
]
def winner(board):
    """
    Returns:
        1  -> X wins
       -1  -> O wins
        0  -> No winner
    """
    for a, b, c in LINES:
        if (
            board[a] != 0
            and board[a] == board[b] == board[c]
        ):
            return board[a]
    return 0
def empty_cells(board):
    """Return all empty cell positions."""
    return [
        i for i in range(9)
        if board[i] == 0
    ]
def place(board, cell, player):
    """Place X or O in a cell."""
    new_board = list(board)
    new_board[cell] = player

    return tuple(new_board)


def random_opponent_move(board):
    """Opponent plays a random legal move."""

    return place(
        board,
        random.choice(empty_cells(board)),
        -1
    )
alpha = 0.3       # Learning rate
gamma = 0.9       # Discount factor
epsilon = 0.2     # Exploration probability
episodes = 50000
Q = defaultdict(float)
visits = defaultdict(int)
def best_move(board):
    """
    Select the empty cell having the highest Q-value.
    """

    return max(
        empty_cells(board),
        key=lambda move: Q[(board, move)]
    )
# ============================================================
# 5. Q-LEARNING TRAINING
# ============================================================

print("Training Q-Learning agent...")

for episode in range(episodes):

    # Empty board
    board = (0,) * 9

    # Sometimes opponent starts
    if random.random() < 0.5:
        board = random_opponent_move(board)

    while True:

        # ----------------------------------------------------
        # Epsilon-greedy strategy
        # ----------------------------------------------------

        if random.random() < epsilon:

            # Explore
            move = random.choice(
                empty_cells(board)
            )

        else:

            # Exploit
            move = best_move(board)

        # Count board visit
        visits[board] += 1

        # AI places X
        next_board = place(
            board,
            move,
            1
        )

        # ----------------------------------------------------
        # Check AI result
        # ----------------------------------------------------

        if winner(next_board) == 1:

            reward = 1.0
            done = True

        elif not empty_cells(next_board):

            reward = 0.5
            done = True

        else:

            # ------------------------------------------------
            # Opponent plays
            # ------------------------------------------------

            next_board = random_opponent_move(
                next_board
            )

            # Opponent wins
            if winner(next_board) == -1:

                reward = -1.0
                done = True

            # Draw
            elif not empty_cells(next_board):

                reward = 0.5
                done = True

            else:

                reward = 0.0
                done = False

        # ----------------------------------------------------
        # Q-Learning update
        # ----------------------------------------------------

        if done:

            target = reward

        else:

            target = (
                reward
                + gamma
                * max(
                    Q[(next_board, m)]
                    for m in empty_cells(next_board)
                )
            )

        # Update Q-value
        Q[(board, move)] += alpha * (
            target - Q[(board, move)]
        )

        # End game
        if done:
            break

        # Move to next state
        board = next_board


print(
    "Training finished."
)

print(
    "Boards seen by agent:",
    len(visits)
)


# ============================================================
# 6. TEST Q-LEARNING AGENT
# ============================================================

def play_one_game():

    board = (0,) * 9

    # Sometimes opponent starts
    if random.random() < 0.5:
        board = random_opponent_move(board)

    while True:

        # AI move
        board = place(
            board,
            best_move(board),
            1
        )

        if winner(board) == 1:
            return "win"

        if not empty_cells(board):
            return "draw"

        # Opponent move
        board = random_opponent_move(board)

        if winner(board) == -1:
            return "loss"

        if not empty_cells(board):
            return "draw"


# Play 1000 games
results = pd.Series(
    [
        play_one_game()
        for _ in range(1000)
    ]
).value_counts()


print("\nQ-Learning results:")
print(results)


# ============================================================
# 7. CREATE DATASET FROM Q-TABLE
# ============================================================

rows = []

# Keep boards seen at least 10 times
for board, count in visits.items():

    if count >= 10:

        rows.append(
            list(board)
            + [best_move(board)]
        )


# Column names
columns = (
    [f"cell_{i}" for i in range(9)]
    + ["best_move"]
)


# Create DataFrame
df = pd.DataFrame(
    rows,
    columns=columns
)


print("\nDataset shape:")
print(df.shape)

print("\nFirst five rows:")
print(df.head())


# ============================================================
# 8. EXPLORE DATASET
# ============================================================

print("\nDataset information:")
print(df.info())

print("\nMissing values:")
print(df.isnull().sum())

print(
    "\nDuplicate rows:",
    df.duplicated().sum()
)

print("\nDataset statistics:")
print(df.describe())


# ============================================================
# 9. VISUALIZE DATASET
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(11, 4)
)


# Distribution of best moves
sns.countplot(
    x="best_move",
    data=df,
    ax=axes[0]
)

axes[0].set_title(
    "Distribution of Best Move"
)


# Number of X and O marks
marks = pd.DataFrame({
    "X marks":
        (df.iloc[:, :9] == 1).sum(axis=1),

    "O marks":
        (df.iloc[:, :9] == -1).sum(axis=1)
})


marks.mean().plot(
    kind="bar",
    ax=axes[1]
)

axes[1].set_title(
    "Average Number of Marks"
)

plt.tight_layout()
plt.show()


# ============================================================
# 10. DATA PREPROCESSING
# ============================================================

# Remove duplicates and missing values
df = df.drop_duplicates().dropna()


# Features
X = df.drop(
    "best_move",
    axis=1
)


# Target
y = df["best_move"]


print("\nFeatures shape:", X.shape)
print("Target shape:", y.shape)


# ============================================================
# 11. TRAIN-TEST SPLIT
# ============================================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.2,
    random_state=SEED,
    stratify=y
)


print(
    "\nTraining samples:",
    X_train.shape[0]
)

print(
    "Testing samples:",
    X_test.shape[0]
)


# ============================================================
# 12. FEATURE SCALING
# ============================================================

scaler = StandardScaler()


# Fit ONLY on training data
X_train_scaled = scaler.fit_transform(
    X_train
)


# Transform test data
X_test_scaled = scaler.transform(
    X_test
)


# ============================================================
# 13. CONVERT TO TENSORFLOW TENSORS
# ============================================================

X_train_tf = tf.convert_to_tensor(
    X_train_scaled,
    dtype=tf.float32
)

X_test_tf = tf.convert_to_tensor(
    X_test_scaled,
    dtype=tf.float32
)

y_train_tf = tf.convert_to_tensor(
    y_train.values,
    dtype=tf.int32
)


print(
    "\nTensor shape:",
    X_train_tf.shape
)

print(
    "Tensor dtype:",
    X_train_tf.dtype
)


# ============================================================
# 14. BUILD NEURAL NETWORK
# ============================================================

def build_model(units):

    model = keras.Sequential([

        # Input = 9 board cells
        layers.Input(shape=(9,)),

        # First hidden layer
        layers.Dense(
            units,
            activation="relu"
        ),

        # Second hidden layer
        layers.Dense(
            units // 2,
            activation="relu"
        ),

        # Output = 9 possible moves
        layers.Dense(
            9,
            activation="softmax"
        )
    ])


    # Compile model
    model.compile(

        optimizer=keras.optimizers.Adam(
            learning_rate=0.001
        ),

        loss="sparse_categorical_crossentropy",

        metrics=["accuracy"]
    )

    return model


# Display model
print("\nModel architecture:")
build_model(64).summary()


# ============================================================
# 15. EARLY STOPPING
# ============================================================

early_stop = keras.callbacks.EarlyStopping(

    monitor="val_loss",

    patience=10,

    restore_best_weights=True
)


# ============================================================
# 16. SELECT BEST HIDDEN LAYER SIZE
# ============================================================

val_scores = {}


for units in [32, 64, 128]:

    print(
        f"\nTraining model with {units} hidden units..."
    )

    tf.keras.utils.set_random_seed(SEED)

    model_temp = build_model(units)

    history_temp = model_temp.fit(

        X_train_tf,

        y_train_tf,

        validation_split=0.2,

        epochs=100,

        batch_size=32,

        callbacks=[early_stop],

        verbose=0
    )


    val_scores[units] = max(
        history_temp.history["val_accuracy"]
    )


print("\nValidation accuracy:")

for units, score in val_scores.items():

    print(
        units,
        "units:",
        round(score, 3)
    )


# Select best model size
best_units = max(
    val_scores,
    key=val_scores.get
)


print(
    "\nSelected hidden layer size:",
    best_units
)


# ============================================================
# 17. TRAIN FINAL MODEL
# ============================================================

tf.keras.utils.set_random_seed(SEED)

model = build_model(
    best_units
)


history = model.fit(

    X_train_tf,

    y_train_tf,

    validation_split=0.2,

    epochs=100,

    batch_size=32,

    callbacks=[early_stop],

    verbose=1
)


print(
    "\nFinal model trained for",
    len(history.history["loss"]),
    "epochs"
)


# ============================================================
# 18. TRAINING CURVES
# ============================================================

fig, axes = plt.subplots(
    1,
    2,
    figsize=(11, 4)
)


# Loss
axes[0].plot(
    history.history["loss"],
    label="Training"
)

axes[0].plot(
    history.history["val_loss"],
    label="Validation"
)

axes[0].set_title("Loss")
axes[0].set_xlabel("Epoch")
axes[0].legend()


# Accuracy
axes[1].plot(
    history.history["accuracy"],
    label="Training"
)

axes[1].plot(
    history.history["val_accuracy"],
    label="Validation"
)

axes[1].set_title("Accuracy")
axes[1].set_xlabel("Epoch")
axes[1].legend()


plt.tight_layout()
plt.show()


# ============================================================
# 19. MODEL PREDICTION
# ============================================================

probabilities = model.predict(
    X_test_tf,
    verbose=0
)


# Choose class with highest probability
y_pred = np.argmax(
    probabilities,
    axis=1
)


print(
    "\nNumber of predictions:",
    len(y_pred)
)

print(
    "\nActual:",
    y_test.values[:10]
)

print(
    "Predicted:",
    y_pred[:10]
)


# ============================================================
# 20. MODEL EVALUATION
# ============================================================

test_loss, test_accuracy = model.evaluate(
    X_test_tf,
    y_test.values,
    verbose=0
)


precision = precision_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


recall = recall_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


f1 = f1_score(
    y_test,
    y_pred,
    average="weighted",
    zero_division=0
)


results_table = pd.DataFrame({

    "Metric": [
        "Test Loss",
        "Accuracy",
        "Precision",
        "Recall",
        "F1-Score"
    ],

    "Value": [
        test_loss,
        test_accuracy,
        precision,
        recall,
        f1
    ]

}).set_index("Metric").round(3)


print("\nModel Evaluation:")
print(results_table)


# ============================================================
# 21. CLASSIFICATION REPORT
# ============================================================

print(
    "\nClassification Report:"
)

print(
    classification_report(
        y_test,
        y_pred,
        zero_division=0
    )
)


# ============================================================
# 22. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(
    y_test,
    y_pred,
    labels=range(9)
)


plt.figure(
    figsize=(6, 5)
)


sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Greys",
    cbar=False
)


plt.title(
    "Confusion Matrix"
)

plt.xlabel(
    "Predicted Move"
)

plt.ylabel(
    "Actual Move"
)

plt.show()


# ============================================================
# 23. USE NEURAL NETWORK TO PLAY
# ============================================================

def network_move(board):

    # Scale board
    x = scaler.transform(
        np.array(board).reshape(1, 9)
    )


    # Convert to tensor
    x = tf.convert_to_tensor(
        x,
        dtype=tf.float32
    )


    # Predict probabilities
    probs = model(
        x,
        training=False
    ).numpy()[0]


    # Remove occupied cells
    occupied = [
        i for i in range(9)
        if board[i] != 0
    ]


    probs[occupied] = -1


    # Select highest probability
    return int(
        np.argmax(probs)
    )


# ============================================================
# 24. PLAY ONE GAME USING NEURAL NETWORK
# ============================================================

def play_one_game_network():

    board = (0,) * 9


    # Sometimes opponent starts
    if random.random() < 0.5:

        board = random_opponent_move(
            board
        )


    while True:

        # Neural network move
        board = place(
            board,
            network_move(board),
            1
        )


        if winner(board) == 1:
            return "win"


        if not empty_cells(board):
            return "draw"


        # Opponent move
        board = random_opponent_move(
            board
        )


        if winner(board) == -1:
            return "loss"


        if not empty_cells(board):
            return "draw"


# ============================================================
# 25. TEST NEURAL NETWORK FOR 1000 GAMES
# ============================================================

network_results = pd.Series(

    [
        play_one_game_network()
        for _ in range(1000)
    ]

).value_counts()


print(
    "\nNeural Network results:"
)

print(
    network_results
)


# ============================================================
# 26. COMPARE Q-LEARNING AND NEURAL NETWORK
# ============================================================

comparison = pd.DataFrame({

    "Q-Learning Agent": results,

    "Neural Network": network_results

}).fillna(0).astype(int)


print(
    "\nComparison:"
)

print(
    comparison
)


# ============================================================
# 27. COMPARISON GRAPH
# ============================================================

comparison.T.plot(
    kind="bar",
    figsize=(7, 4),
    colormap="gray",
    edgecolor="black"
)


plt.title(
    "1000 Games Against Random Opponent"
)

plt.ylabel(
    "Number of Games"
)

plt.xticks(
    rotation=0
)

plt.show()


# ============================================================
# 28. FINAL RESULTS
# ============================================================

print("\n====================================")
print("FINAL RESULTS")
print("====================================")


print(
    "Test Accuracy:",
    round(
        results_table.loc[
            "Accuracy",
            "Value"
        ],
        3
    )
)
print(
    "Test F1-Score:",
    round(
        results_table.loc[
            "F1-Score",
            "Value"
        ],
        3
    )
)
print(
    "Final Training Accuracy:",
    round(
        history.history["accuracy"][-1],
        3
    )
)
print(
    "Final Validation Accuracy:",
    round(
        history.history["val_accuracy"][-1],
        3
    )
)
q_wins = results.get("win", 0)
nn_wins = network_results.get("win", 0)
print(
    "\nQ-Learning Win Rate:",
    q_wins / 10,
    "%"
)
print(
    "Neural Network Win Rate:",
    nn_wins / 10,
    "%"
)
```
